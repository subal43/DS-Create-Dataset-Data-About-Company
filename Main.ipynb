```python
from bs4 import BeautifulSoup
import pandas as pd
import requests
```

```python
webpage = requests.get("https://www.ambitionbox.com/list-of-companies?page=2",headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/94.0.4606.81 Safari/537.36',
    'Accept-Language': 'en-US,en;q=0.9',
    'Referer': 'https://www.google.com/' # optional, helps mimic browsing history
}
).text
```

```python
bsoup = BeautifulSoup(webpage , "lxml")
```

```python
company =  bsoup.find_all('div', class_="companyCardWrapper")
```

```python
cmp_name = []
ratings = []
reviews = []
ctype = []
hq_cmp = []
for i in company :
    cmp_name.append(i.find('h2', class_="companyCardWrapper__companyName").text.strip())
    ratings.append(i.find('div', class_="rating_text rating_text--md").text.strip())
    reviews.append(i.find('span', class_="companyCardWrapper__companyRatingCount").text.strip().replace("(", "").replace(")", ""))
    ctype.append(i.find('span', class_="companyCardWrapper__interLinking").text.strip().split("|")[0])
    hq_cmp.append(i.find('span', class_="companyCardWrapper__interLinking").text.strip().split("|")[1].split(" ")[1])
```

```python
main_data = []
for page in range(1,11):
    webpage = requests.get(f"https://www.ambitionbox.com/list-of-companies?page={page}",headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/94.0.4606.81 Safari/537.36',
    'Accept-Language': 'en-US,en;q=0.9',
    'Referer': 'https://www.google.com/' # optional, helps mimic browsing history
    }
    ).text

    bsoup = BeautifulSoup(webpage , "lxml")
    company =  bsoup.find_all('div', class_="companyCardWrapper")

    cmp_name = []
    ratings = []
    reviews = []
    ctype = []
    hq_cmp = []
    for i in company :
        cmp_name.append(i.find('h2', class_="companyCardWrapper__companyName").text.strip())
        ratings.append(i.find('div', class_="rating_text rating_text--md").text.strip())
        reviews.append(i.find('span', class_="companyCardWrapper__companyRatingCount").text.strip().replace("(", "").replace(")", ""))
        ctype.append(i.find('span', class_="companyCardWrapper__interLinking").text.strip().split("|")[0])
        hq_cmp.append(i.find('span', class_="companyCardWrapper__interLinking").text.strip().split("|")[1].split(" ")[1])
        d = {"Company_Name":cmp_name , "Ratings":ratings , "Reviews":reviews , "company_type":ctype , "Headquarters":hq_cmp}
    df = pd.DataFrame(d)
    main_data.append(df)
final_df = pd.concat(main_data, ignore_index=True)

print(final_df)

  
        
```
